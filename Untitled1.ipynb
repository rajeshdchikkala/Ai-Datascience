```python
import pandas as pd
import numpy as np
from scipy import stats

df =pd.read_csv("/content/mtcars.csv")

df.head()
```

```python
#Ttest
#null hypothesis -avergae  milleage of cars =20mpg
#aleternative hypothesis - average mileage is not equal to 20

#A manufacturer claims that the average mileage of cars is 20 mpg. Test whether the mtcars dataset supports this claim.
#if p value <= 0.05 then we accept alternate hypothesis
#if p value >= 0.05 then we accept null hypothesis

t_stat,p_val = stats.ttest_1samp(df['mpg'],20)
print("t_stat:",t_stat)
print("p_val:",p_val)
```

```python
#A average horsepower =150
t_stat,p_val = stats.ttest_1samp(df['hp'],150)
print("t_stat:",t_stat)
print("p_val:",p_val)
```

```python
#The average weight of cars is believed to be 3.0 (1000 lbs). Test whether the dataset supports this claim.
t_stat,p_val = stats.ttest_1samp(df['wt'],3.00)
print("t_stat:",t_stat)
print("p_val:",p_val)
```

```python
t_stat,p_val = stats.ttest_1samp(df['qsec'],18)
print("t_stat:",t_stat)
print("p_val:",p_val)
```

```python
#mpg difference (manual vs automatic)
#null hypo : if mpg is same
#alt null : mpg is not same

auto = df[df['am']==1]['mpg']
manual = df[df['am']==0]['mpg']

t_stat,p_val = stats.ttest_ind(auto,manual)
print("t_stat:",t_stat)
print("p_val:",p_val)
```

```python
#HP diff 4 cyl vs 6cyl
cyl4 = df[df['cyl']==4]['hp']
cyl6 = df[df['cyl']==6]['hp']

t_stat,p_val = stats.ttest_ind(cyl4,cyl6)
print("t_stat:",t_stat)
print("p_val:",p_val)
```

```python
#is transmissions related to cylinder count
#chi square test
contingency = pd.crosstab(df['am'],df['cyl'])
contingency
```

```python
chi2,p_val,dof,expected = stats.chi2_contingency(contingency)
print("chi2:",chi2)
print("p_val:",p_val)
print("dof:",dof) # degree of freedom
print("expected:",expected)
```

```python
#mpg category vs transmission
df['mpg_category'] = pd.cut(df['mpg'],bins=2,labels=['low','high'])
contingency = pd.crosstab(df['am'],df['mpg_category'])
contingency

chi2,p,dof,expected = stats.chi2_contingency(contingency)
print("chi2:",chi2)
print("p_val:",p_val)
print("dof:",dof) # degree of freedom
print("expected:",expected)
```

```python
#ANOVA TEST MORE THAN TWO DIFFERENT CATEGORIES
#does mpg differ across cylinder group
#independent test 2 categories with a continuous
#Anova we compare more 2 categories with continuous
import  statsmodels.api as sm
from statsmodels.formula.api import ols

model = ols('mpg~C(cyl)',data=df).fit()
anova_table = sm.stats.anova_lm(model,typ=2)
print(anova_table)
```

```python
#Hp across gear types
model = ols('hp~C(gear)',data=df).fit()
anova_table = sm.stats.anova_lm(model,typ=2)
print(anova_table)
```

```python
import pandas as pd
data = {
    "TV": [50, 60, 70, 80, 90, 100, 110],
    "Radio": [30, 35, 40, 45, 50, 55, 60],
    "Social_Media": [20, 25, 30, 35, 40, 45, 50],
    "Sales": [120, 140, 150, 170, 180, 200, 210]
}
df = pd.DataFrame(data)
print(df)
```

```python
from sklearn.linear_model import LinearRegression
X=df[['TV','Radio','Social_Media']]
y=df['Sales']
model = LinearRegression()
model.fit(X,y)
```

```python
b0 = model.intercept_
b1,b2,b3 = model.coef_
print("Intercept (b0):",b0)
print("Tv coefficient(b1):",b1)
print("Radio coefficient(b2):",b2)
print("Social Media Coefficient(b3):",b3)
```

```python
new_data =[[120,65,55]]
predicted_sales = model.predict(new_data)
print("Predicted Sales:",predicted_sales[0])
```

```python
new_data =[[80,45,35]]
predicted_sales = model.predict(new_data)
print("Predicted Sales:",predicted_sales[0])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
y_pred = model.predict(X)
mae = mean_absolute_error(y, y_pred)
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y, y_pred)
print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R2 Score:", r2)
```
