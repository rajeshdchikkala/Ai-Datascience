```python
def area_rect(l,b):
    return l*b
l= int(input("entern the length:"))
b= int(input("entern the breadth:"))
area = area_rect(l,b)
print("Area of rectangle:", area)
```

```python
def area_circle(l):
    return 3.14 *l*l
l= int(input("entern the radius:"))
area = area_circle(l)
print("Area of circle:", area)
```

```python
def find_positive(number):
    if number > 0:
       return "positive"
    elif number == 0:
      return "neither positive nor negative"
    else:
      return "negative"
l= int(input("Enter the number:"))
print("numebr is:", find_positive(l))
```
